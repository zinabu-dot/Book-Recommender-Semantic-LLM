## Zero-shot classification from Hugging Face

Zero-Shot Classification (ZSC) is a machine learning technique that enables models to classify data into categories they have never encountered during training. Unlike traditional supervised learning, which requires labeled examples for every class, ZSC leverages prior knowledge to generalize to unseen classes. This is particularly useful in scenarios where collecting labeled data for every possible category is impractical or impossible.

#### How It Works

Zero-shot classification relies on semantic knowledge transfer to bridge the gap between seen and unseen classes. This is achieved using techniques such as word embeddings (e.g., `Word2Vec`, `GloVe`), pre-trained language models (e.g., `BERT`, `GPT`, `CLIP`), or knowledge graphs. The model maps both input data and class descriptions into a shared embedding space and uses similarity measures (e.g., `cosine` similarity) to determine the most relevant class.

For example, if a model trained on "cats" and "dogs" is asked to classify a "lion," it can infer that a lion is semantically similar to a cat and make the correct classification.

#### Key Components

1. **Semantic Embeddings:** Classes and input data are represented as vector embeddings that capture their meaning.

2. **Similarity Matching:** The model compares the input's embedding with class embeddings and selects the class with the highest similarity score.

```python
from transformers import pipeline
# Initialize the zero-shot classification pipeline
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli", device="cpu")

candidate_labels = ["Fiction", "Nonfiction"]
```

```python
# Perform classification
description_seq = books.loc[books['new_categories']=="Fiction", "description"].iloc[0]

result = classifier(description_seq, candidate_labels)
```

```python
import numpy as np
def category_predictor(text_sequence: str, proposed_catagories: list[str]):
    """ Obtain the predicted probability of each category. Pick the category with the highest probability. """
    result = classifier(text_sequence, proposed_catagories)
    idx = np.argmax(result['scores'])
    predicted_label = result['labels'][idx]
    return predicted_label
```

# Evaluate this model

First we take a sample of 300 books from each group.

Next, we make a prediction of categories for each book description.

Lastly, we compare the actual groups with predicted ones, and thus evaluate the model's ability to generalize making categories, so that we use it to predict the nan ones in the 'new_category' column.

```python
fiction_sample = books[books['new_categories'] == 'Fiction'].sample(300, random_state=42)
nonfiction_sample = books[books['new_categories'] == 'Nonfiction'].sample(300, random_state=42)

result_df = pd.concat([fiction_sample, nonfiction_sample], ignore_index=True)
```

```python
actual_categories = []
predicted_categories = []
for _, row in result.iterrows():
    predicted_label = category_predictor(row['description'], candidate_labels)
    actual_categories.append(row['new_categories'])
    predicted_categories.append(predicted_label)
```

```python
from sklearn.metrics import classification_report
print(classification_report(actual_categories, predicted_categories))
```

```python
from tqdm import tqdm
missing_cats = books.loc[books['new_categories'].isna(), ['isbn13', 'description']]
PRED_CATS = []
for _, row in tqdm(missing_cats.iterrows(), total=len(missing_cats)):
    predicted_label = category_predictor(row['description'], candidate_labels)
    data = {
        'isbn13': row['isbn13'],
        'new_categories': predicted_label
    }
    PRED_CATS.append(data)
```

```python
predicted_cats_df = pd.DataFrame(PRED_CATS).rename(columns={'new_categories': 'pred_new_categories'})
predicted_cats_df
```

```python
books_df = pd.merge(books, predicted_cats_df, on='isbn13', how='left')
books_df['new_categories'] = np.where(books_df['new_categories'].isna(),books_df['pred_new_categories'], books_df['new_categories'])
books_final = books_df.drop(columns="pred_new_categories").copy()
books_final
```
